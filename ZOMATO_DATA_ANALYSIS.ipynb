```python
import pandas as pd
```

```python
data=pd.read_csv("/content/Zomato-data-.csv",delimiter=',')
data
```

```python
data.shape
```

```python
data.head()
data.tail()
data.info()
data.describe()
```

```python
data.isnull()
data.isnull().sum()
```

```python

data['rate']=pd.to_numeric(data['rate'],errors='coerce')
data.describe()
```

```python
import matplotlib.pyplot as plt
restaurant_type = data['listed_in(type)'].value_counts()

plt.bar(restaurant_type.index, restaurant_type.values)
plt.title('Number of Restaurants by Type')
plt.xlabel('Restaurant Type')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
restaurant_type = data['listed_in(type)'].value_counts()

plt.pie(restaurant_type.values, labels=restaurant_type.index, autopct='%1.1f%%')
plt.title('Restaurant')
plt.show()
```

```python
plt.hist(data['rate'], bins=10)
plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
import seaborn as sb
heatmap_data = data[['rate','votes','approx_cost(for two people)']]
sb.heatmap(heatmap_data)
plt.title('Zomato Data Heatmap')
plt.xlabel('Features')
plt.ylabel('Restaurants')
plt.show()
```
